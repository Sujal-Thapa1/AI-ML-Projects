# My First Ai Agent

```python
import random

class SimpleAIAgent:
    def __init__(self):
        self.responses = {
            "hello": ["Hi there! How can I help you?", "Hey! What's up?", "Hello! Nice to meet you."],
            "how are you": ["I'm doing great, thanks for asking!", "I'm functioning perfectly!", "All systems operational!"],
            "what is your name": ["I'm SimpleBot, your AI assistant!", "You can call me SimpleBot!", "I'm SimpleBot, nice to know you!"],
            "help": ["I can chat with you! Just ask me anything.", "I'm here to have a conversation with you.", "Feel free to ask me questions!"],
            "bye": ["Goodbye! Have a great day!", "See you later!", "Catch you soon!"],
        }
    
    def respond(self, user_input):
        user_input = user_input.lower().strip()
        
        for key in self.responses:
            if key in user_input:
                return random.choice(self.responses[key])
        
        return "I'm not sure about that. Can you ask something else?"
    
    def chat(self):
        print("SimpleBot: Hello! I'm SimpleBot. Type 'bye' to exit.")
        while True:
            user_input = input("You: ")
            if user_input.lower() == "bye":
                print("SimpleBot:", random.choice(self.responses["bye"]))
                break
            response = self.respond(user_input)
            print(f"SimpleBot: {response}")

# Run the chatbot
agent = SimpleAIAgent()
agent.chat()
```
